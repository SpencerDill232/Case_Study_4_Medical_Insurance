```python
# write here
import pandas as pd

# GitHub raw URL for the medical insurance dataset
github_url = "https://raw.githubusercontent.com/SpencerDill232/Case_Study_4_Medical_Insurance/main/medical_insurance.csv"

# Load the dataset directly using pandas
df = pd.read_csv(github_url)

# Display basic information about the dataset
if df is not None:
    print("\nDataset Info:")
    print(df.info())
    print("\nFirst few rows:")
    print(df.head())
    print("\nBasic statistics:")
    print(df.describe())

# 3. Check the target distribution for is_high_risk
print("\n'is_high_risk' value counts:")
print(df["is_high_risk"].value_counts())

print("\n'is_high_risk' proportion:")
print(df["is_high_risk"].value_counts(normalize=True))
```

```python
# =============================================================================
# CORRELATION ANALYSIS WITH RISK SCORE
# =============================================================================

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Define the variables of interest
variables_of_interest = [
    'hypertension', 'diabetes', 'copd', 'cardiovascular_disease', 'cancer_history',
    'kidney_disease', 'liver_disease', 'arthritis', 'mental_health',
    'chronic_count', 'systolic_bp', 'diastolic_bp', 'ldl', 'hba1c',
    'visits_last_year', 'hospitalizations_last_3yrs', 'days_hospitalized_last_3yrs',
    'medication_count', 'proc_imaging_count', 'proc_surgery_count', 'proc_physio_count',
    'proc_consult_count', 'proc_lab_count', 'had_major_procedure'
]

# Add risk_score to the list for correlation analysis
correlation_vars = variables_of_interest + ['risk_score']

# Select only the columns that exist in the dataset
available_vars = [var for var in correlation_vars if var in df.columns]
missing_vars = [var for var in correlation_vars if var not in df.columns]

if missing_vars:
    print(f"\nWarning: The following variables were not found in the dataset:")
    print(missing_vars)

# Create correlation matrix
correlation_matrix = df[available_vars].corr()

# Extract correlations with risk_score and sort by absolute value
risk_correlations = correlation_matrix['risk_score'].drop('risk_score').sort_values(
    key=abs, ascending=False
)

# Display correlation table
print("\n" + "="*70)
print("CORRELATION WITH RISK SCORE (sorted by impact)")
print("="*70)
print(f"{'Variable':<35} {'Correlation':<15} {'Abs Value':<15}")
print("-"*70)

for var, corr in risk_correlations.items():
    print(f"{var:<35} {corr:>14.4f} {abs(corr):>14.4f}")

print("="*70)

# Create visualization of correlations with risk_score
plt.figure(figsize=(10, 8))
risk_correlations.plot(kind='barh', color=['red' if x < 0 else 'green' for x in risk_correlations])
plt.xlabel('Correlation Coefficient')
plt.title('Correlation of Variables with Risk Score', fontsize=14, fontweight='bold')
plt.axvline(x=0, color='black', linestyle='-', linewidth=0.5)
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

# Create full correlation heatmap
plt.figure(figsize=(16, 14))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0,
    square=True,
    linewidths=0.5,
    cbar_kws={'label': 'Correlation Coefficient'}
)
plt.title('Full Correlation Matrix - All Variables', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

# Summary statistics
print("\n" + "="*70)
print("CORRELATION SUMMARY STATISTICS")
print("="*70)
print(f"Strongest positive correlation: {risk_correlations.max():.4f} ({risk_correlations.idxmax()})")
print(f"Strongest negative correlation: {risk_correlations.min():.4f} ({risk_correlations.idxmin()})")
print(f"Mean absolute correlation: {risk_correlations.abs().mean():.4f}")
print(f"Median absolute correlation: {risk_correlations.abs().median():.4f}")
print("="*70)
```

```python
# =============================================================================
# CORRELATION ANALYSIS WITH PRICE VARIABLES
# =============================================================================

print("\n\n" + "="*70)
print("PRICE VARIABLES CORRELATION ANALYSIS")
print("="*70)

# Define price variables and analysis variables
price_variables = [
    'annual_medical_cost',
    'annual_premium',
    'monthly_premium',
    'avg_claim_amount',
    'total_claims_paid'
]

analysis_variables = [
    'risk_score', 'is_high_risk', 'visits_last_year', 'hospitalizations_last_3yrs',
    'days_hospitalized_last_3yrs', 'medication_count', 'proc_imaging_count',
    'proc_surgery_count', 'proc_physio_count', 'proc_consult_count',
    'proc_lab_count', 'had_major_procedure',
    'deductible', 'copay', 'policy_term_years', 'policy_changes_last_2yrs',
    'provider_quality'
]

# Combine all variables for this analysis
price_analysis_vars = analysis_variables + price_variables

# Select only columns that exist
available_price_vars = [var for var in price_analysis_vars if var in df.columns]
missing_price_vars = [var for var in price_analysis_vars if var not in df.columns]

if missing_price_vars:
    print(f"\nWarning: The following variables were not found in the dataset:")
    print(missing_price_vars)

# Create correlation matrix for price analysis
price_correlation_matrix = df[available_price_vars].corr()

# For each price variable, show correlations with all other variables
for price_var in price_variables:
    if price_var in df.columns:
        print(f"\n{'='*70}")
        print(f"CORRELATIONS WITH {price_var.upper()}")
        print("="*70)

        # Get correlations with this price variable, excluding other price variables
        price_corrs = price_correlation_matrix[price_var].drop(
            [p for p in price_variables if p in price_correlation_matrix.columns]
        ).sort_values(key=abs, ascending=False)

        print(f"{'Variable':<40} {'Correlation':<15} {'Abs Value':<15}")
        print("-"*70)

        for var, corr in price_corrs.items():
            print(f"{var:<40} {corr:>14.4f} {abs(corr):>14.4f}")

        # Summary for this price variable
        print("-"*70)
        print(f"Strongest positive: {price_corrs.max():.4f} ({price_corrs.idxmax()})")
        print(f"Strongest negative: {price_corrs.min():.4f} ({price_corrs.idxmin()})")
        print(f"Mean absolute: {price_corrs.abs().mean():.4f}")

# Create visualization comparing all price variables
fig, axes = plt.subplots(len([p for p in price_variables if p in df.columns]), 1,
                         figsize=(12, 5*len([p for p in price_variables if p in df.columns])))

if len([p for p in price_variables if p in df.columns]) == 1:
    axes = [axes]

plot_idx = 0
for price_var in price_variables:
    if price_var in df.columns:
        # Exclude other price variables from correlations
        price_corrs = price_correlation_matrix[price_var].drop(
            [p for p in price_variables if p in price_correlation_matrix.columns]
        ).sort_values(key=abs, ascending=False).head(15)  # Show top 15 correlations

        price_corrs.plot(
            kind='barh',
            ax=axes[plot_idx],
            color=['red' if x < 0 else 'green' for x in price_corrs]
        )
        axes[plot_idx].set_xlabel('Correlation Coefficient', fontsize=11)
        axes[plot_idx].set_title(f'Top 15 Correlations with {price_var}',
                                fontsize=12, fontweight='bold')
        axes[plot_idx].axvline(x=0, color='black', linestyle='-', linewidth=0.5)
        axes[plot_idx].grid(axis='x', alpha=0.3)
        plot_idx += 1

plt.tight_layout()
plt.show()

# Create heatmap showing correlations between price variables and key predictors
# Select top correlated variables with any price variable for the heatmap
all_price_corrs = pd.DataFrame()
for price_var in price_variables:
    if price_var in df.columns:
        # Exclude other price variables
        corrs_without_prices = price_correlation_matrix[price_var].drop(
            [p for p in price_variables if p in price_correlation_matrix.columns]
        )
        all_price_corrs[price_var] = corrs_without_prices

# Get variables with highest average absolute correlation across all price variables
avg_abs_corr = all_price_corrs.abs().mean(axis=1).sort_values(ascending=False).head(20)
top_vars = avg_abs_corr.index.tolist()

# Create focused heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(
    price_correlation_matrix.loc[top_vars, [p for p in price_variables if p in df.columns]],
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0,
    linewidths=0.5,
    cbar_kws={'label': 'Correlation Coefficient'}
)
plt.title('Top 20 Variables Correlated with Price Variables', fontsize=14, fontweight='bold')
plt.xlabel('Price Variables', fontsize=12)
plt.ylabel('Predictor Variables', fontsize=12)
plt.tight_layout()
plt.show()

print("\n" + "="*70)
print("PRICE CORRELATION ANALYSIS COMPLETE")
print("="*70)
```

```python
# =============================================================================
# CATEGORICAL VARIABLES ANALYSIS WITH PRICE VARIABLES
# =============================================================================

print("\n\n" + "="*70)
print("CATEGORICAL VARIABLES vs PRICE VARIABLES ANALYSIS")
print("="*70)

from sklearn.preprocessing import LabelEncoder

# Define categorical variables
categorical_variables = [
    'plan_type', 'network_tier', 'alcohol_freq', 'smoker',
    'employment_status', 'marital_status', 'education',
    'urban_rural', 'region', 'sex'
]

# Check which categorical variables exist in the dataset
available_categorical = [var for var in categorical_variables if var in df.columns]
missing_categorical = [var for var in categorical_variables if var not in df.columns]

if missing_categorical:
    print(f"\nWarning: The following categorical variables were not found:")
    print(missing_categorical)

print(f"\nFound {len(available_categorical)} categorical variables to analyze")
print(f"Categorical variables: {', '.join(available_categorical)}")

# Create a copy of the dataframe for encoding
df_encoded = df.copy()

# Label encode categorical variables
label_encoders = {}
for cat_var in available_categorical:
    if cat_var in df_encoded.columns:
        le = LabelEncoder()
        # Handle missing values
        df_encoded[cat_var] = df_encoded[cat_var].fillna('Unknown')
        df_encoded[f'{cat_var}_encoded'] = le.fit_transform(df_encoded[cat_var].astype(str))
        label_encoders[cat_var] = le

# Get encoded column names
encoded_cols = [f'{var}_encoded' for var in available_categorical]

# Combine encoded categoricals with price variables for correlation
categorical_price_vars = encoded_cols + [p for p in price_variables if p in df_encoded.columns]
categorical_corr_matrix = df_encoded[categorical_price_vars].corr()

# Analysis for each price variable
for price_var in price_variables:
    if price_var in df_encoded.columns:
        print(f"\n{'='*70}")
        print(f"CATEGORICAL CORRELATIONS WITH {price_var.upper()}")
        print("="*70)

        # Get correlations with categorical variables only
        cat_corrs = categorical_corr_matrix[price_var][encoded_cols].copy()
        # Remove the '_encoded' suffix for display
        cat_corrs.index = [idx.replace('_encoded', '') for idx in cat_corrs.index]
        cat_corrs = cat_corrs.sort_values(key=abs, ascending=False)

        print(f"{'Categorical Variable':<30} {'Correlation':<15} {'Abs Value':<15}")
        print("-"*70)

        for var, corr in cat_corrs.items():
            print(f"{var:<30} {corr:>14.4f} {abs(corr):>14.4f}")

        print("-"*70)
        print(f"Strongest positive: {cat_corrs.max():.4f} ({cat_corrs.idxmax()})")
        print(f"Strongest negative: {cat_corrs.min():.4f} ({cat_corrs.idxmin()})")
        print(f"Mean absolute: {cat_corrs.abs().mean():.4f}")

# Visualization: Categorical correlations with each price variable
available_prices = [p for p in price_variables if p in df_encoded.columns]
fig, axes = plt.subplots(len(available_prices), 1,
                         figsize=(12, 5*len(available_prices)))

if len(available_prices) == 1:
    axes = [axes]

for idx, price_var in enumerate(available_prices):
    cat_corrs = categorical_corr_matrix[price_var][encoded_cols].copy()
    cat_corrs.index = [i.replace('_encoded', '') for i in cat_corrs.index]
    cat_corrs = cat_corrs.sort_values(key=abs, ascending=False)

    cat_corrs.plot(
        kind='barh',
        ax=axes[idx],
        color=['red' if x < 0 else 'green' for x in cat_corrs]
    )
    axes[idx].set_xlabel('Correlation Coefficient', fontsize=11)
    axes[idx].set_title(f'Categorical Variable Correlations with {price_var}',
                       fontsize=12, fontweight='bold')
    axes[idx].axvline(x=0, color='black', linestyle='-', linewidth=0.5)
    axes[idx].grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.show()

# Create comprehensive heatmap
plt.figure(figsize=(10, 8))
cat_price_heatmap = categorical_corr_matrix.loc[
    encoded_cols,
    available_prices
].copy()
# Clean up row labels
cat_price_heatmap.index = [idx.replace('_encoded', '') for idx in cat_price_heatmap.index]

sns.heatmap(
    cat_price_heatmap,
    annot=True,
    fmt='.3f',
    cmap='coolwarm',
    center=0,
    linewidths=0.5,
    cbar_kws={'label': 'Correlation Coefficient'}
)
plt.title('Categorical Variables vs Price Variables - Correlation Heatmap',
          fontsize=14, fontweight='bold')
plt.xlabel('Price Variables', fontsize=12)
plt.ylabel('Categorical Variables', fontsize=12)
plt.tight_layout()
plt.show()

# Show category counts and mean prices for better interpretation
print("\n" + "="*70)
print("CATEGORY STATISTICS - Mean Values by Category")
print("="*70)

for cat_var in available_categorical:
    if cat_var in df.columns:
        print(f"\n{cat_var.upper()}:")
        print("-"*70)

        # For each price variable, show mean by category
        for price_var in price_variables:
            if price_var in df.columns:
                mean_by_category = df.groupby(cat_var)[price_var].agg(['mean', 'count'])
                mean_by_category = mean_by_category.sort_values('mean', ascending=False)

                print(f"\n  {price_var}:")
                print(f"  {'Category':<25} {'Mean':<15} {'Count':<10}")
                for category, row in mean_by_category.iterrows():
                    print(f"  {str(category):<25} ${row['mean']:>13,.2f} {int(row['count']):>9}")

print("\n" + "="*70)
print("CATEGORICAL ANALYSIS COMPLETE")
print("="*70)
```
